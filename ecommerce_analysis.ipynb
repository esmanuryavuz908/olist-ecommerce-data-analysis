```python
def teslimat_durumu_belirle(gecikme):
    if pd.isna(gecikme):          
        return "Teslim edilmedi"
    elif gecikme < 0:             
        return "Erken teslim edildi"
    elif gecikme == 0:            
        return "Zamanında teslim edildi"
    elif 1 <= gecikme <= 3:       
        return "1–3 gün gecikti"
    elif 4 <= gecikme <= 7:       
        return "4–7 gün gecikti"
    else:                     
        return "7 günden fazla gecikti"
        
ana_tablo["teslimat_durumu"] = ana_tablo["gecikme_gun"].apply(teslimat_durumu_belirle)
ana_tablo[["gecikme_gun", "teslimat_durumu"]].sample(4)
```

```python
eyalet_analizi = ana_tablo.groupby("customer_state").agg(
    Musteri_Sayisi=("customer_id", "nunique"),          
    Siparis_Sayisi=("order_id", "nunique"),             
    Toplam_Gelir=("total_amount", "sum"),                
    Ortalama_Kargo_Ucreti=("freight_value", "mean"),    
    Ortalama_Teslimat_Suresi=("teslimat_suresi_gun", "mean"),
    Ortalama_Musteri_Puani=("review_score", "mean")    
)
eyalet_analizi["Ortalama_Siparis_Tutari"] = eyalet_analizi["Toplam_Gelir"] / eyalet_analizi["Siparis_Sayisi"]
eyalet_analizi = eyalet_analizi.sort_values(by="Toplam_Gelir", ascending=False)
eyalet_analizi.head()
```

```python
satici_analizi = ana_tablo.groupby("seller_id").agg(
    Toplam_Siparis_Sayisi=("order_id", "nunique"),           
    Satilan_Urun_Sayisi=("product_id", "count"),             
    Toplam_Gelir=("total_amount", "sum"),                    
    Ortalama_Urun_Fiyati=("price", "mean"),                  
    Ortalama_Kargo_Ucreti=("freight_value", "mean"),        
    Ortalama_Musteri_Puani=("review_score", "mean"),        
    Ortalama_Teslimat_Suresi=("teslimat_suresi_gun", "mean"), 
    Geciken_Siparis_Sayisi=("is_delayed", "sum")             
)
satici_analizi["Gecikme_Orani"] = satici_analizi["Geciken_Siparis_Sayisi"] / satici_analizi["Satilan_Urun_Sayisi"]
satici_analizi.sample(5)
```

```python
def satici_siniflandir(satir):
    puan = satir["Ortalama_Musteri_Puani"]
    gecikme = satir["Gecikme_Orani"]
    siparis = satir["Toplam_Siparis_Sayisi"]
    
    if siparis < 10:
        return "Yeni/Az Satış Yapan"
    elif puan >= 4.0 and gecikme <= 0.10:
        return "Çok başarılı"
    elif puan >= 3.5 and gecikme <= 0.20:
        return "Başarılı"
    elif puan < 3.0 or gecikme >= 0.40:
        return "Riskli"
    else:
        return "Ortalama"

satici_analizi["Performans_Sinifi"] = satici_analizi.apply(satici_siniflandir, axis=1)
print(satici_analizi["Performans_Sinifi"].value_counts())
```

```python
odeme_analizi = ana_tablo.groupby("payment_type").agg(
    Kullanim_Sayisi=("payment_type", "count"),          
    Toplam_Odeme=("payment_value", "sum"),             
    Ortalama_Odeme=("payment_value", "mean"),           
    Ortalama_Taksit=("payment_installments", "mean"),   
    En_Yuksek_Odeme=("payment_value", "max"),           
    En_Dusuk_Odeme=("payment_value", "min")             
)
odeme_analizi = odeme_analizi.sort_values(by="Kullanim_Sayisi", ascending=False)
odeme_analizi.round(2)
```

```python
print("--- ÖDEME YÖNTEMLERİ RAPORU ---\n")
en_cok_kullanilan = odeme_analizi["Kullanim_Sayisi"].idxmax()
print(f"1. En Fazla Kullanılan Ödeme Yöntemi: {en_cok_kullanilan}")
en_cok_gelir_getiren = odeme_analizi["Toplam_Odeme"].idxmax()
print(f"2. En Yüksek Gelir Getiren Yöntem: {en_cok_gelir_getiren}")
kredi_karti_taksit = odeme_analizi.loc["credit_card", "Ortalama_Taksit"]
print(f"3. Kredi Kartı Ortalama Taksit Sayısı: {kredi_karti_taksit:.1f}")
taksit_fiyat_iliskisi = ana_tablo.groupby("payment_installments")["payment_value"].mean()
print("\n4. Taksit Sayısına Göre Ortalama Ödeme Tutarları:")
print(taksit_fiyat_iliskisi.head(6).round(1))
print("(Yorum: Listeye bakarak taksit sayısı ile fiyatın paralel artıp artmadığını görebiliriz.)")
tek_cekim_ortalama = ana_tablo[ana_tablo["payment_installments"] == 1]["payment_value"].mean()
taksitli_ortalama = ana_tablo[ana_tablo["payment_installments"] > 1]["payment_value"].mean()
print(f"\n5. Tek Çekim Ortalama Tutar: {tek_cekim_ortalama:.2f}")
print(f"   Taksitli Ortalama Tutar: {taksitli_ortalama:.2f}")
```

```python
def memnuniyet_siniflandir(puan):
    if pd.isna(puan):          
        return "Bilinmiyor"
    elif puan <= 2:               
        return "Memnun değil"
    elif puan == 3:               
        return "Kararsız"
    elif puan >= 4:               
        return "Memnun"
        
ana_tablo["memnuniyet_grubu"] = ana_tablo["review_score"].apply(memnuniyet_siniflandir)
memnuniyet_analizi = ana_tablo.groupby("memnuniyet_grubu").agg(
    Siparis_Sayisi=("order_id", "nunique"),                 
    Ortalama_Urun_Fiyati=("price", "mean"),                   
    Ortalama_Kargo_Ucreti=("freight_value", "mean"),          
    Ortalama_Teslimat_Suresi=("teslimat_suresi_gun", "mean"), 
    Ortalama_Gecikme_Suresi=("gecikme_gun", "mean")           
)
memnuniyet_analizi = memnuniyet_analizi.sort_values(by="Siparis_Sayisi", ascending=False)
print(memnuniyet_analizi.round(1))
```

```python
print("--- MÜŞTERİ MEMNUNİYETİ VE GECİKME ANALİZİ ---\n")
gecikenler_puan_ort = ana_tablo[ana_tablo["gecikme_gun"] > 0]["review_score"].mean()
zamaninda_puan_ort = ana_tablo[ana_tablo["gecikme_gun"] <= 0]["review_score"].mean()
print(f"1. Geciken siparişlerin ortalama puanı: {gecikenler_puan_ort:.2f}")
print(f"2. Zamanında/Erken teslim edilenlerin ortalama puanı: {zamaninda_puan_ort:.2f}")
print("\n3. Veri Yorumu (Korelasyon):")
print("Sayılar gösteriyor ki; teslimat geciktiğinde ortalama müşteri puanında belirgin bir düşüş var.")
print("Ancak kesin bir 'neden-sonuç' kuramayız. Belki de geciken kargolar aynı zamanda hasarlı gittiği için puan düştü. Biz sadece aralarındaki negatif ilişkiyi (gecikme artar, puan düşer) gözlemliyoruz.")

print("\n--- KATEGORİ VE SATICI ANALİZİ ---\n")
ana_tablo["is_low_score"] = (ana_tablo["review_score"] <= 2).astype(int)
dusuk_puanli_tablo = ana_tablo[ana_tablo["is_low_score"] == 1]
kategori_kotu_puanlar = dusuk_puanli_tablo["product_category_name_english"].value_counts().head(5)
print("4. En fazla düşük puan (1 ve 2) alan ilk 5 kategori:")
print(kategori_kotu_puanlar)
satici_puan_analizi = ana_tablo.groupby("seller_id").agg(
    Toplam_Satis=("order_id", "count"),
    Kotu_Puan_Sayisi=("is_low_score", "sum")
)
satici_puan_analizi["Kotu_Puan_Orani"] = satici_puan_analizi["Kotu_Puan_Sayisi"] / satici_puan_analizi["Toplam_Satis"]
guvenilir_saticilar = satici_puan_analizi[satici_puan_analizi["Toplam_Satis"] >= 30]
en_riskli_saticilar = guvenilir_saticilar.sort_values(by="Kotu_Puan_Orani", ascending=False).head(5)
print("\n5. Düşük puan alma ORANI en yüksek olan 5 satıcı (En az 30 satış yapanlar):")
print(en_riskli_saticilar[["Toplam_Satis", "Kotu_Puan_Orani"]].round(2))
```

```python
ana_tablo['order_purchase_timestamp'] = pd.to_datetime(ana_tablo['order_purchase_timestamp'])
ana_tablo['Siparis_Ayi'] = ana_tablo['order_purchase_timestamp'].dt.month
ocak_verisi = ana_tablo[ana_tablo['Siparis_Ayi'] == 1]
subat_verisi = ana_tablo[ana_tablo['Siparis_Ayi'] == 2]
mart_verisi = ana_tablo[ana_tablo['Siparis_Ayi'] == 3]

def aylik_ozet_cikar(df, ay_adi):
    ozet_tablo = pd.DataFrame({
        "Ay": [ay_adi],
        "Sipariş_sayısı": [df['order_id'].nunique()],
        "Toplam_gelir": [df['total_amount'].sum()],
        "Ortalama_tutar": [df['total_amount'].sum() / df['order_id'].nunique()],
        "Ortalama_puan": [df['review_score'].mean()]
    })
    return ozet_tablo
ocak_ozet = aylik_ozet_cikar(ocak_verisi, "Ocak")
subat_ozet = aylik_ozet_cikar(subat_verisi, "Şubat")
mart_ozet = aylik_ozet_cikar(mart_verisi, "Mart")
print("--- OCAK AYI ÖZET TABLOSU ---")
print(ocak_ozet.round(2))
```

```python
yillik_rapor = pd.concat([ocak_ozet, subat_ozet, mart_ozet], ignore_index=True)
print("--- BİRLEŞTİRİLMİŞ YILLIK RAPOR ---")
print(yillik_rapor.round(2))
```

```python
toplam_siparis = ana_tablo['order_id'].nunique()
toplam_musteri = ana_tablo['customer_id'].nunique()
toplam_satici = ana_tablo['seller_id'].nunique()
toplam_urun = ana_tablo['product_id'].nunique()
toplam_gelir = ana_tablo['payment_value'].sum()
toplam_kargo = ana_tablo['freight_value'].sum()
ort_siparis_tutari = toplam_gelir / toplam_siparis
ort_teslimat_suresi = ana_tablo['teslimat_suresi_gun'].mean()
ort_puan = ana_tablo['review_score'].mean()
geciken_siparis_sayisi = ana_tablo[ana_tablo['gecikme_gun'] > 0]['order_id'].nunique()
gecikme_orani = geciken_siparis_sayisi / toplam_siparis
en_kategori = ana_tablo['product_category_name_english'].value_counts().index[0]
en_satici = ana_tablo['seller_id'].value_counts().index[0]
en_eyalet = ana_tablo['customer_state'].value_counts().index[0]
en_odeme = ana_tablo['payment_type'].value_counts().index[0]

sirket_ozeti = pd.DataFrame({
    "Toplam_Sipariş": [toplam_siparis],
    "Toplam_Müşteri": [toplam_musteri],
    "Toplam_Satıcı": [toplam_satici],
    "Toplam_Ürün": [toplam_urun],
    "Toplam_Gelir": [f"{toplam_gelir:,.2f} ₺"], 
    "Toplam_Kargo_Geliri": [f"{toplam_kargo:,.2f} ₺"],
    "Ortalama_Sipariş_Tutarı": [f"{ort_siparis_tutari:,.2f} ₺"],
    "Ort_Teslimat_Süresi(Gün)": [round(ort_teslimat_suresi, 1)],
    "Gecikme_Oranı": [f"%{gecikme_orani * 100:.1f}"], 
    "Ortalama_Puan": [round(ort_puan, 2)],
    "En_Başarılı_Kategori": [en_kategori],
    "En_Başarılı_Satıcı": [en_satici],
    "En_Çok_Satış_Yapılan_Eyalet": [en_eyalet],
    "En_Çok_Kullanılan_Ödeme": [en_odeme]
})

print("--- YÖNETİM ÖZET TABLOSU ---")
print(sirket_ozeti.T)
```

```python
category_report = ana_tablo.groupby('product_category_name_english').agg(
    Siparis_Sayisi=('order_id', 'nunique'),
    Ortalama_Fiyat=('price', 'mean'),
    Ortalama_Puan=('review_score', 'mean')
).sort_values(by='Siparis_Sayisi', ascending=False)

state_report = ana_tablo.groupby('customer_state').agg(
    Musteri_Sayisi=('customer_id', 'nunique'),
    Siparis_Sayisi=('order_id', 'nunique'),
    Toplam_Gelir=('payment_value', 'sum')
).sort_values(by='Toplam_Gelir', ascending=False)

seller_report = ana_tablo.groupby('seller_id').agg(
    Satilan_Urun_Sayisi=('product_id', 'count'), 
    Toplam_Kazanilan=('payment_value', 'sum'),
    Ortalama_Puan=('review_score', 'mean')
).sort_values(by='Satilan_Urun_Sayisi', ascending=False)

payment_report = ana_tablo.groupby('payment_type').agg(
    Kullanim_Sayisi=('order_id', 'nunique'),
    Toplam_Hacim=('payment_value', 'sum')
).sort_values(by='Kullanim_Sayisi', ascending=False)

ana_tablo['Teslimat_Durumu'] = ana_tablo['gecikme_gun'].apply(lambda x: "Gecikti" if x > 0 else "Zamanında/Erken")
delivery_report = ana_tablo.groupby('Teslimat_Durumu').agg(
    Siparis_Sayisi=('order_id', 'nunique'),
    Ortalama_Teslimat_Suresi=('teslimat_suresi_gun', 'mean'),
    Ortalama_Puan=('review_score', 'mean')
)

monthly_report = ana_tablo.groupby('Siparis_Ayi').agg(
    Siparis_Sayisi=('order_id', 'nunique'),
    Toplam_Gelir=('payment_value', 'sum'),
    Ortalama_Siparis_Tutari=('payment_value', 'mean')
).sort_index() 

print("--- TÜM DETAY RAPORLARI BAŞARIYLA OLUŞTURULDU ---")
print("Raporları incelemek için yeni hücrelerde 'category_report.head()' gibi komutlar yazabilirsin.")
```
